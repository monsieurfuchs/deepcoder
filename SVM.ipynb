# Support Vector Machine

Die lineare und polynomiale Regression, die wir im vorherigen Abschnitt kennengelernt haben, dienten dazu aus einer reihe von Werten neue Werte vorherzusagen. Eine andere Gruppe als die der Regressionsverfahren sind die Klassifikationsverfahren, bei denen es darum geht, die Zugehörigkeit zu einer Klasse bzw. Kategorie voherzusagen - also etwa "Prüfung bestanden" versus "Prüfung nicht bestanden" oder "kreditwürdig" versus "nicht kreditwürdig".

Ein sehr bekanntes Verfahren zur Trennung zweier Klassen und der entsprechenden Vorhersage ist die Support Vector Machine (SVM). 
Eine SVM teilt Objekte in Klassen, indem sie er eine **Trennebene** (in höheren Dimensionen eine d-dimensionale Hyperebene) konstruiert, welche den Abstand zu den nächsten Samples jeder Klasse maximiert. Abbildung 1 veranschaulicht, wie eine SVM zwei Klasen voneinander trennt. Die SVM findet jene bestmögliche Trennebene zwischen den beiden Klassen, bei denen die äußersten Samples der beiden Klassen jeweils denselben Abstand zur Trennebene haben. Gleichzeitig soll dieser Abstand - **Margin** genannt - maximal groß sein. Besagte äußerste Punkte werden **Stützvektoren** bzw. **Support Vectors** genannt.

## Modelldefinition

Das Vorgehen entspricht dem üblichen Ablauf eines maschinellen Lernverfahrens:

1. Erstellung des mathematischen Modells bestehend aus Gewichten und Bias
2. Erstellung einer zu minimierenden Kostenfunktion
3. Bildung der partiellen ersten Ableitungen der Kostenfunktion
4. Ermittlung der Modellparameter mittels Gradientenabstieg
5. Vorhersage unbekannter Werte mit den ermittelten Modellparametern

Unser mathematisches Modell für die Trennebene entspricht der allgemeinen Gleichung einer Ebene mit den Gewichten $w_1, w_2,\dots,w_d$, wobei $d$ die Anzahl der Dimensionen ist und die Gewichte sich zum Gewichtsvektor $w$ zusammenfassen lassen. Die Gleichung der gesuchten Ebene lautet:

$$w^Tx + b = 0 \tag{1}$$

Der Abstand eines Punktes $x_0$ zu dieser Ebene ist gegeben durch:

$$d = \frac{|w^Tx_0 + b|}{||w||} \tag{2}$$

Hierbei ist $||w||$ die Norm des Gewichtsvektors $w$.

## Konstruktion der Kostenfunktion

Wie erwähnt enthält $w$ die **Gewichte** - $b$ ist hingegen der **Bias** unseres Modells. Es gilt nun, $w$ und $b$ so zu bestimmen dass erstens der Abstand zwischen den Klassen maximal wird und zweitens möglichst wenig Fehlklassifikationen entstehen. Unser Ziel ist es, hierfür eine Kostenfunktion zu finden.

### Stützvektoren und Stützebenen

Die Menge aller möglichen Stützvektoren können wir uns als zwei weitere Ebenen. denken, die in einem bestimmten, jeweils gleichen Abstand parallel zur Trennebene positioniert sind. Wir nennen diese zwei Ebenen an dieser Stelle **Stützebenen** und definieren sie durch folgende zwei Gleichungen:

$$S^{+}: w^Tx^{+} + b = 1 \tag{3}$$
$$S^{-}: w^Tx^{-} + b = -1 \tag{4}$$

Die Versatzwerte $1$ und $-1$ sind willkürlich gewählt. Die tatsächliche Lage der beiden Ebenen und somit ihr jeweiliger Abstand zur Trennebene sind durch die Parameter $w$ und $b$ gegeben, die es zu bestimmen gilt.

Nehmen wir an, unsere Daten unterteilen sich in zwei Klassen: jene Personen, die eine Prüfung bestanden haben und jene, die sie nicht bestanden haben. Erstere zählen wir als **positive Samples**, zweitere als **negative Samples**. 
Alle positiven samples sind gegeben durch:

$$w^Tx^{+} + b \geq 1 \tag{5}$$

Alle negativen Samples sind gegeben durch:

$$w^Tx^{-} + b \geq 1 \tag{6}$$

### Abstand zwischen den Stützebenen

Uns interessiert der Abstand zwischen den beiden Stützebenen, denn eben diesen wollen wir maximieren. Denken wir uns einen beliebigen Stützvektor $x^{-}$ für ein negatives Sample und einen ebenso beliebigen Stützvektor $x^{+}$ für ein positives Sample. $x^{+}$ liegt also auf der Ebene $S^{+}$, wohingegen $x^{-}$ auf der Ebene $S^{-}$ liegt.

Den Abstand zwischen den beiden Stützebenen erhalten wir entsprechend Abbildung xxx durch folgendes Skalarprodukt:

$$d = (x^{+} - x^{-})^T\frac{w}{||w||} \tag{7}$$

Dies ergibt sich aus dem Umstand, dass das Skalarprodukt eines Vektors mit einem Einheitsvektor ($w/||w||$) seine Projektion entlang des Einheitsvektors ergibt. Da $w$ senkrecht zur Trennebene und damit auch senkrecht zu den Stützebenen liegt, entspricht dies dem Abstand zwischen den Stützebenen. Nun wollen wir versuchen, $x_{+}$ und $x_{-}$ in dieser Gleichung aufzulösen. Hierzu stellen wir die Gleichung 3 und 4 etwas um, wir werden sogleich sehen warum.

$$S^{+}: w^Tx^{+} + b = 1 \Longleftrightarrow w^Tx^{+} = 1 - b  \tag{8}$$
$$S^{-}: w^Tx^{-} + b = -1 \Longleftrightarrow -w^Tx^{-} = 1 + b \tag{9}$$

Nun begeben wir uns zurück zu Gleichung 7:

$$d = (x^{+} - x^{-})^T \frac{w}{||w||} \tag{10}$$
$$ = \frac{w^T(x^{+} - x^{-})}{||w||} \tag{11}$$
$$ = \frac{w^Tx^{+} - w^Tx^{-}}{||w||} \tag{12}$$

Hier können wir für $w^Tx^{+}$ und $-w^Tx^{-}$ die Terme aus den Gleichungen 8 und 9 einsetzen:

$$d = \frac{1 - b + 1 + b}{||w||} \tag{13}$$
$$ = \frac{2}{||w||} \tag{14}$$

Wir haben nun eine erstaunlich simple Formel für die Berechnung des Abstands zwischen den Stützebenen hergeleitet. Genau diesen ABstand wollen wir maximieren. Erinnern wir uns: das allgemeine Rezept für ein maschinelles Lernverfahren beinhaltet die Ermiztlung einer Kostenfunktion und ihre anschließende Optimierung mittels des Gradientenabstiegs. Die Maximierung des Abstabnds $d$ ist reziprok zur Minimerung seines Kehrwerts:

$$min\left(\frac{||w||}{2}\right) \tag{15}$$

Um das Minimum zu finden, benötigen wir die Nullstelle der ersten Ableitung. Damit die Ableitung einfacher wird, können wir das Quadrat der Norm $||w||$ bilden. DIes ist nur ein mathematischer Kniff, um uns das Leben an dieser Stelle einfacher zu machen.

$$min\left(\frac{||w||^2}{2}\right) \tag{16}$$

### Hinge-Loss-Funktion
Die Maximierung des Margins $d$ ist jedoch nicht unser einziges Optimierungsziel. Ebenso soll der Algorithmus möglichst wenig Fehlklassifikationen liefern. Es gilt somit, die SVM während des Trainings für falsch vorhergesagte Samples zu "bestrafen". Hierfür nutzen wir die **Hinge-Loss-Funktion**, welche häufig bei Klassifikationsverfahren zum Einsatz kommt:

$$L(y_i, f(x_i)) = max(0, 1 - y_i \cdot f(x_i)) \tag{17}$$

Hierbei ist $x_i$ der $i$te Sample-Wert und $y_i$ die vorhergesagte Klasse:

$$y_i = \left\{\begin{array}{ll} -1  & \text{für } x_- \text{ (negative Samples)} \\ 1 & \text{für } x_+ \text{ (positive Samples)}\end{array}\right. \tag{18}$$

$f(x_i)$ ist dabei die Funktion der Hyperebene, die unsere Klassen trennt und für alle Punkte auf dieser Ebene $0$ ergibt (siehe hierzu Gleichung 1):

$$w^Tx_i + b = 0 \tag{19}$$

$$L(y_i, f(x_i)) = max(0, 1 - y_i \cdot (w^Tx_i + b))) \tag{20}$$

Welchen genauen Effekt hat diese Form der Hinge-Loss-Funktion und wie "bestraft" sie das Training für falsche Klassifikationen? Erinnern wir uns, dass für unsere beiden Margin-Grenzen das Folgende gilt:

$$w^Tx_+ + b \geq 1 \text{ für positive Samples} \tag{21}$$
$$w^Tx_- + b \leq 1 \text{ für negative Samples} \tag{22}$$

Generell sind zwei Gleichungen zur Beschreibung ein und desselben Sachverhaltes unhandlich. Um daraus eine Gleichung zu bilden, können wir zu einem Trick greifen. Wir hatten für die Klassenwerte $y_i$ die Werte $1$ für positive Samples $x_+$ und $-1$ für negative Samples $x_-$ gewählt. Wenn wir die beiden Gleichungen mit den entsprechenen Werten multiplizieren erhalten wir:

$$y_i(w^Tx + b) \geq 1 \text{ für } y_i = 1 \text{ und } x = x_+ \tag{22}$$
$$y_i(w^Tx + b) \geq 1 \text{ für } y_i = -1 \text{ und } x = x_- \tag{23}$$

Beide Gleichungen sind nun dieselben und wir erkennen, dass der linke Term dem Ausdruck in unserer Hinge-Loss-Funktion entspricht. 

Dieser Term ist $\geq 1$, wenn das klassifizierte Sample außerhalb des Margins oder unmittelbar darauf liegt und zudem der richtigen Klasse zugeordnet wurde. Der Wert der Hinge-Loss-Funktion ist dann $0$.

Sofern ein Sample hingegen innerhalb des Margins liegt oder falsch klassifiziert wurde, nimmt der Term einen Wert $ \lt 1$ an und die Hinge-Funktion wird $\gt 0$. 

Nun führen wir unsere beiden Minimierungsprobleme zu einer Kostenfunktion zusammen:

$$L(w, b) = \frac{1}{2} ||w||^2 + C \sum_{i=1}^{m} max(0, 1 - y_i((w^Tx_i +b)) $$

Unsere Kostenfunktion kombiniert nun also eine Maximirung des Margin-Raums mit einem Bestrafungswert (Penalty) für falsch klassifizierte Samples. Der Parameter $C$ ist dabei ein Regularisierungsfaktor, der darüber bestimmt, wie stark die jeweiligen Kostenanteile relativ zueinander gewichtet werden. 

### Ableitung der Kostenfunktion
Um den Gradientanabstieg durchzuführen und so das Minimum der Kostenfunktion zu finden, müssen wir diese partiell jeweils nach $w$ und $b$ ableiten. Die Ableitung gestaltet aufgrund der nicht stetigen Hinge-Loss-Funktion etwas komplizierter. Wir müssen hierbei mehrere Fälle voneinander unterscheiden:

#### Fall 1
Ein Sample wird richtig klassifiziert und befindet sich außerhalb des Margins oder auf einem seiner Ränder. Dann wird der Hinge Loss $0$ und wir müssen nur die folgende Funktion ableiten:
$$L(w,b) = \frac{1}{2}||w||^2$$
$w$ ist ein Vektor mit den Komponenten $(w_1, w_2, w_3,\dots)$. Darum gilt:
$$ \frac{\partial L}{\partial w} = \frac{\partial}{\partial w} \frac{||w||^2}{2}$$
$$\text{ mit } ||w|| = (w_1^2 + w_2^2 +  w_3^2 + \dots)$$
Da $w$ ein Vektor ist, müssen wir die partielle Ableitung ebenfalls vektoriell, also für jede Komponente $w_j$ durchführen:
$$\frac{\partial L}{\partial w} = \begin{pmatrix} 
\frac{\partial }{\partial w_1}\left(\frac{1}{2}(w_1^2 + w_2^2 +  w_3^2 + \dots)\right) \\ 
\frac{\partial }{\partial w_2}\left(\frac{1}{2}(w_1^2 + w_2^2 +  w_3^2 + \dots))\right) \\ 
\frac{\partial }{\partial w_3}\left(\frac{1}{2}(w_1^2 + w_2^2 +  w_3^2 + \dots)\right) \\
\vdots
\end{pmatrix} =
\begin{pmatrix}
w_1 \\
w_2 \\
w_3 \\
\vdots
\end{pmatrix} = w$$
Wir erhalten also für den Fall, dass Samples richtig klassifiziert werden und innerhalb des Margins oder auf seinen Rändern liegen, die folgenden beiden partiellen Ableitungen:

$$\frac{\partial L}{\partial w} = w $$
$$\frac{\partial L}{\partial b} = 0$$
